```python
# 1. Given a string count the number of vowels and return the count

def count_vowels(string):
    vowels = "aeiouAEIOU"
    count = 0
    for char in string:
        if char in vowels:
            count += 1
    return count
s = input("Enter a word or a sentence:")
print("Number of vowels:", count_vowels(s))
```

```python
# 2. Give a string count the number of each vowel in the string and return it as a dictionary

def count_each_vowel(string):
    vowels = "aeiou"
    vowel_count = {
        "a": 0,
        "e": 0,
        "i": 0,
        "o": 0,
        "u": 0
    }
    for char in string.lower():
        if char in vowels:
            vowel_count[char] += 1
    return vowel_count
s =input("Enter a word or a sentence:")
print(count_each_vowel(s))
```

```python
# 3. Write a function to check if a number is prime or not.

def prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True
number = int(input("Enter a number:"))
if prime(number):
    print("Prime")
else:
    print("Not Prime")
```

```python

# 4. Write a Python program to print even-length words in a string.

def even_length_words(string):
    words = string.split()
    for word in words:
        if len(word) % 2 == 0:
            print(word)
x = input("Enter a word or a sentence:")
even_length_words(x)
```

```python
# 5. Write a Python function  to print a given sequence

def print_pattern(n):
    for i in range(1, n + 1):
        for j in range(i):
            print(i, end=" ")
        print()
print_pattern(3)
```

```python
# 6. Write a Python program to check whether a string is symmetrical or palindrome.

def check_string(string):
    if string == string[::-1]:
        print("The string is a Palindrome")
    else:
        print("The string is not a Palindrome")
    length = len(string)
    if length % 2 == 0:
        first_half = string[:length // 2]
        second_half = string[length // 2:]
        if first_half == second_half:
            print("The string is Symmetrical")
        else:
            print("The string is not Symmetrical")
    else:
        print("The string is not Symmetrical")
string = input("Enter a word or a sentence:")
check_string(string)
```

```python
# 8. Given a list of numbers, return the sum of the elements inside the list.

def list_sum(numbers):
    total = 0
    for number in numbers:
        total += number
    return total
numbers = eval(input("Enter the list: "))
print("Sum:", list_sum(numbers))
```

```python
# 9. Create a class named as “operations” where each method is an arithmetic operation

class operations:
    def __init__(self, number):
        self.number = number
    def addition(self, value):
        return self.number + value
    def subtraction(self, value):
        return self.number - value
    def multiplication(self, value):
        return self.number * value
    def division(self, value):
        if value == 0:
            return "Cannot divide by zero"
        return self.number / value
obj = operations(20)
print("Addition:", obj.addition(10))
print("Subtraction:", obj.subtraction(10))
print("Multiplication:", obj.multiplication(10))
print("Division:", obj.division(10))
```

```python
# Given a list of numbers, sort it into two different lists of even and odd numbers separately.
def separate_even_odd(numbers):
    even_numbers = []
    odd_numbers = []
    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)
        else:
            odd_numbers.append(number)
    return even_numbers, odd_numbers
numbers = eval(input("Enter the list of numbers: "))
even, odd = separate_even_odd(numbers)
print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
# 11. Given a dictionary of marks
#marks = { "math": 80, "english": 75, "science": 90, "computer": 95 }
#Return the subject name that has maximum marks.

def maximum_marks(marks):
    maximum_subject = max(marks, key=marks.get)
    return maximum_subject
marks = {
    "math": 80,
    "english": 75,
    "science": 90,
    "computer": 95
}
print("Subject with maximum marks:", maximum_marks(marks))
```

```python
# 12. Given a list of numbers with duplicates, return True if there is at least one duplicate in the list.

def has_duplicate(numbers):
    seen = set()
    for number in numbers:
        if number in seen:
            return True

        seen.add(number)
    return False
numbers = eval(input("Enter the list of numbers: "))

print(has_duplicate(numbers))
```

```python
# 13. Given two strings, check whether they are anagram or not.

def is_anagram(string1, string2):
    string1 = string1.replace(" ", "").lower()
    string2 = string2.replace(" ", "").lower()

    return sorted(string1) == sorted(string2)
string1 = input("enter a word")
string2 = input("enter another word")

if is_anagram(string1, string2):
    print("The strings are Anagrams")
else:
    print("The strings are not Anagrams")
```

```python
# 14. Best Time to Buy and Sell Stock – Leetcode problem

def max_profit(prices):
    if not prices:
        return 0
    minimum_price = prices[0]
    maximum_profit = 0
    for price in prices:
        if price < minimum_price:
            minimum_price = price

        profit = price - minimum_price
        if profit > maximum_profit:
            maximum_profit = profit
    return maximum_profit
prices = [7, 1, 5, 3, 6, 4]
print("Maximum Profit:", max_profit(prices))
```

```python
# 15. Given a list of numbers nums = [100,4,200,1,3,2] find the longest consecutive sequence.

def longest_consecutive(nums):
    numbers = set(nums)
    longest = 0
    for number in numbers:
        if number - 1 not in numbers:
            current = number
            length = 1
            while current + 1 in numbers:
                current += 1
                length += 1
            longest = max(longest, length)
    return longest
nums = eval(input("Enter the list: "))
print("Longest consecutive sequence:", longest_consecutive(nums))
```
